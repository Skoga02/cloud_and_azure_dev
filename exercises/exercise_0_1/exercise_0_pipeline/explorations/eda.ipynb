```python
import matplotlib.pyplot as plt

df = spark.table("airbnb.hosts.airbnb_bronze")

reviews_pd = df.select("number_of_reviews").dropna().toPandas()

fig, ax = plt.subplots(figsize=(12, 5))

ax.hist(reviews_pd["number_of_reviews"], bins=50, color="#378ADD", edgecolor="white", linewidth=0.5)

ax.set_xlabel("Number of reviews")
ax.set_ylabel("Number of accomomodations")
ax.set_title("Distribution of number of reviews")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
from pyspark.sql.functions import col, regexp_replace, coalesce, when, lit

df = spark.read.table("airbnb.hosts.airbnb_bronze")

df_clean = (
    df
    .withColumn(
        "price_clean",
        regexp_replace(col("price"), "[$,]", "").cast("double")
    )
    .withColumn(
        "number_of_reviews",
        col("number_of_reviews").cast("int")
    )
    .withColumn(
        "review_rate_number",
        col("review_rate_number").cast("double")
    )
    .withColumn(
        "last_review",
        coalesce(col("last_review").cast("string"), lit("missing"))
    )
    .withColumn(
        "reviews_per_month",
        coalesce(col("reviews_per_month").cast("string"), lit("missing"))
    )
    .withColumn(
        "house_rules",
        coalesce(col("house_rules").cast("string"), lit("missing"))
    )
    .withColumn(
        "license",
        coalesce(col("license").cast("string"), lit("missing"))
    )
).drop(
    "_rescued_data"
)

df_clean.select("last_review").display()
```
